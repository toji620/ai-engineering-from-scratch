# Exercise 1: list comprehension vs numpy

Create 100,000 random numbers two ways and time each with `%timeit`, which runs the line many times and reports the average.

```python
import random
import numpy as np

N = 100_000
```

## Pure Python: a list comprehension
Calls `random.random()` once per number, in a Python loop.

```python
%timeit [random.random() for _ in range(N)]
```

## NumPy: one call
`np.random.rand(N)` fills the whole array in compiled C.

```python
%timeit np.random.rand(N)
```
